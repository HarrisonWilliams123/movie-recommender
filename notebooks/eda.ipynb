```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Opens the csv file and prints out the first five entries
df = pd.read_csv(r"C:/Users/harri/Documents/movie-recommender/data/netflix_titles.csv")
df.head()
```

```python
#Prints out the Null values of the datset (Will use an imputer later, keep Null for now)
df.isnull().sum()
```

```python
#Checks to see how many duplicated rows there are
df.duplicated().sum()
```

```python
#Shows the value counts to see if it's a movie or tv show
df['type'].value_counts(normalize=True)
```

```python
#Makes a graph of how many movies vs. tv shows there are
counts = df['type'].value_counts()
plt.figure(figsize=(10,6))
plt.bar(counts.index, counts.values, color='skyblue')
plt.xlabel("Category")
plt.ylabel("Count")
plt.title("Frequency of Categories")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
#Bar Graph to show where most tv shows and movies come from
counts = df['country'].value_counts().head(10)
plt.figure(figsize=(10,6))
plt.bar(counts.index, counts.values, color='skyblue')
plt.xlabel("Countries")
plt.ylabel("Count")
plt.title("Frequency of Countries")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
